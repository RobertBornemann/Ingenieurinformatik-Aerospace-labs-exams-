```python
import math
import sympy as sp
```

Find the magnitude and direction of a vector sum

```python
def vector_x_y(magnitude,direction): 
    direction = math.radians(direction)
    x_value = magnitude*math.cos(direction)
    y_value = magnitude*math.sin(direction)
    return [x_value, y_value]

def sum_of_vectors(vector_a, vector_b):
    return [
        vector_a[0] + vector_b[0],
        vector_a[1] + vector_b[1]
    ]

def magnitude_x_y(vector_sum):
    magnitude_vector_sum = math.sqrt(vector_sum[0]**2+vector_sum[1]**2)
    return magnitude_vector_sum
    
def direction_x_y(vector_sum):
    direction_vector_sum = math.degrees(
        math.atan2(vector_sum[1], vector_sum[0])
    )

    if direction_vector_sum < 0:
        direction_vector_sum += 360

    return direction_vector_sum
    
```

```python
vector_a_mag = 10
vector_a_dir = 188
vector_b_mag = 10
vector_b_dir = 352

vector_a = vector_x_y(vector_a_mag,vector_a_dir)
vector_b = vector_x_y(vector_b_mag,vector_b_dir)

total_sum_of_vectors = sum_of_vectors(vector_a, vector_b)

total_magnitude = magnitude_x_y(total_sum_of_vectors)
total_direction = direction_x_y(total_sum_of_vectors)

print(round(total_magnitude, 3), round(total_direction, 3))

```

Find the magnitude of a vector scalar multiple

```python
def magnitude_scalar(vector, scalar):
    scalar_magnitude = abs(scalar)*sp.sqrt(vector[0]**2 + vector[1]**2)
    return scalar_magnitude
```

```python
vector = [-9,9]
scalar = 9

result = magnitude_scalar(vector, scalar)
print(result)

```

Find a unit vector

```python
def unit_vector(vector):
    unit_vector_magnitude = sp.sqrt(vector[0]**2 + vector[1]**2)
    divide_unit_vector = [
        vector[0] / unit_vector_magnitude,
        vector[1] / unit_vector_magnitude
    ]
    #confirm_unit_vector = sp.sqrt(divide_unit_vector[0]**2 + divide_unit_vector[1]**2)
    return divide_unit_vector
    #return confirm_unit_vector
```

```python
v = [-5,-8]

result = unit_vector(v)
print(result)
```

 Find the magnitude of a three-dimensional vector

```python
def magnitude_x_y_z(vector_a, vector_b):
    vector_sum = [
        vector_b[0] - vector_a[0],
        vector_b[1] - vector_a[1],
        vector_b[2] - vector_a[2]
    ]
    magnitude_vector_sum = sp.sqrt(vector_sum[0]**2+vector_sum[1]**2+vector_sum[2]**2)
    #magnitude_vector_sum = math.sqrt(vector_sum[0]**2+vector_sum[1]**2+vector_sum[2]**2)
    return magnitude_vector_sum
```

```python
v1 = [6,2,-10]
v2 = [-4,9,-2]

result = magnitude_x_y_z(v1,v2)
print(result)
```
