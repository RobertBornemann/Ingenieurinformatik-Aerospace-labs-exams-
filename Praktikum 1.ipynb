```python
batterie_type = 'Sony SE US26650VT 2600mAh (Green)' 
kapazitaet_mAh = 2600
spannung_V = 3.7
masse_g = 88
strom_mA = 50
```

```python
print(f'Kapazität (mAh): {type(kapazitaet_mAh)} \n',
        f'Spannung (V): {type(spannung_V)} \n',
        f'Masse (g): {type(masse_g)} \n',
        f'Strom (mA): {type(strom_mA)} \n',

     )
```

```python
laufzeit_h = kapazitaet_mAh/strom_mA
energieinhalt_Wh = spannung_V*kapazitaet_mAh*1e-3
spezifische_energie_Wh_pro_Kg = energieinhalt_Wh/(masse_g*1e-3)

print(f'Laufzeit: {laufzeit_h} h, {type(laufzeit_h)} \n',
        f'Energieinhalt: {energieinhalt_Wh} Wh, {type(energieinhalt_Wh)} \n',
        f'Spezifische Energie: {round(spezifische_energie_Wh_pro_Kg,2)} Wh/Kg, {type(spezifische_energie_Wh_pro_Kg)} \n'
     )
```

```python
zellenzahl_pack = 96
pack_spannung = zellenzahl_pack*spannung_V
energieinhalt = pack_spannung*(kapazitaet_mAh*1e-6)

print(round(energieinhalt,2))
```
